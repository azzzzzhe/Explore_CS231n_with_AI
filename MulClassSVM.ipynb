```python
class MultiClassSVM(nn.Module):
    def __init__(self,delta=1.0):
        super(MultiClassSVM, self).__init__()
        self.delta = delta

    def forward(self,scores,target):
        return MultiClassSVM_loss(scores,target,self.delta)

def MultiClassSVM_loss(X, y,delta=1.0):
    """Multi-class SVM loss function. X is matrix of shape (N, C)"""
    N = X.shape[0]
    
    # 获取正确类别的得分，形状 (N, 1)
    correct_scores = X[range(N), y].reshape(-1, 1)
    
    # 计算边际: max(0, s_j - s_yi + delta)
    margins = X - correct_scores + delta
    
    # 使用 torch.maximum 替代 torch.max（逐元素比较）
    margins = torch.maximum(margins, torch.tensor(0.0))
    
    # 将正确类别的边际设为 0（不参与损失计算）
    margins[range(N), y] = 0
    
    # 数据损失
    loss = torch.sum(margins) / N
    
    return loss
```
